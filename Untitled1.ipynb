```python
import torch
import torch.nn.functional as F
from torchvision import models, transforms
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

# 1) Load and preprocess the image
img_path = '/content/1-s2.0-S0263931909001811-gr3.jpg'  # put the JPEG in your working dir
orig = Image.open(img_path).convert('RGB')
prep = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225])
])
input_tensor = prep(orig).unsqueeze(0)

# 2) Load a lightweight pretrained model
model = models.mobilenet_v2(pretrained=True)
model.eval()

# 3) Grad-CAM implementation
class GradCAM:
    def __init__(self, model, target_layer):
        self.model = model
        self.gradients = None
        self.activations = None
        # register hooks
        target_layer.register_forward_hook(
            lambda m, i, o: setattr(self, 'activations', o))
        target_layer.register_backward_hook(
            lambda m, gi, go: setattr(self, 'gradients', go[0]))

    def __call__(self, x, class_idx=None):
        logits = self.model(x)
        if class_idx is None:
            class_idx = logits.argmax(dim=1).item()
        self.model.zero_grad()
        logits[0, class_idx].backward()
        grads = self.gradients[0]      # [C, H, W]
        acts  = self.activations[0]    # [C, H, W]
        weights = grads.mean(dim=(1,2))  # [C]
        cam = torch.relu((weights[:, None, None] * acts).sum(dim=0))
        cam = (cam - cam.min())/(cam.max() - cam.min())
        return cam.detach().cpu().numpy(), class_idx

# 4) Hook into the last convolutional block of MobileNetV2
grad_cam = GradCAM(model, model.features[-1])

# 5) Generate the map
cam, pred = grad_cam(input_tensor)
heatmap = np.uint8(255 * cam)
heatmap = Image.fromarray(heatmap).resize(orig.size, Image.LANCZOS)
heatmap_np = np.array(heatmap)

# 6) Create overlay
overlay = np.array(orig) * 0.6 + np.stack([heatmap_np]*3, axis=2) * 0.4

# 7) Display
fig, ax = plt.subplots(1, 3, figsize=(15, 5))
ax[0].imshow(orig);      ax[0].set_title("Original");      ax[0].axis('off')
ax[1].imshow(heatmap_np, cmap='jet'); ax[1].set_title("Grad‑CAM"); ax[1].axis('off')
ax[2].imshow(np.uint8(overlay));     ax[2].set_title(f"Overlay (class {pred})"); ax[2].axis('off')
plt.tight_layout()
plt.show()
```

```python
import torch
import torchxrayvision as xrv
import numpy as np
import matplotlib.pyplot as plt
from torchvision import transforms
from PIL import Image

# 1) Load a pretrained CheXNet (DenseNet121) model
model = xrv.models.DenseNet(weights="chexnet")
model.eval()

# 2) Preprocess: xrv expects a 1‑channel, normalized tensor
orig = Image.open("1-s2.0-S0263931909001811-gr3.jpg").convert("L")
prep = transforms.Compose([
    transforms.Resize((320, 320)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485], std=[0.229])
])
input_t = prep(orig).unsqueeze(0)  # [1,1,320,320]

# 3) Grad‑CAM helper (same idea, but pointing to DenseNet’s last conv)
class GradCAM:
    def __init__(self, model, layer):
        self.model = model
        self.activations = None
        self.gradients   = None
        layer.register_forward_hook(
            lambda m, i, o: setattr(self, "activations", o))
        layer.register_backward_hook(
            lambda m, gi, go: setattr(self, "gradients",   go[0]))

    def __call__(self, x, class_idx):
        logits = self.model(x)
        loss   = logits[0, class_idx]
        self.model.zero_grad()
        loss.backward()
        grads = self.gradients[0]
        acts  = self.activations[0]
        weights = grads.mean(dim=(1,2))
        cam = torch.relu((weights[:,None,None] * acts).sum(dim=0))
        cam = (cam - cam.min()) / (cam.max() - cam.min())
        return cam.cpu().numpy()

# 4) Hook to the very last DenseNet feature map
grad_cam = GradCAM(model, model.features.denseblock4.denselayer16)

# 5) Find the pneumonia class index in CheXNet’s outputs
classes = model.pathologies  # ['Atelectasis', ..., 'Pneumonia', ...]
pneum_idx = classes.index("Pneumonia")

# 6) Compute and visualize
cam_map = grad_cam(input_t, pneumonia_idx)
heat = np.uint8(255 * cam_map)
heat = Image.fromarray(heat).resize(orig.size, Image.LANCZOS)
heat_np = np.array(heat)
overlay = np.stack([orig, orig, orig], axis=2)*0.6 + np.stack([heat_np]*3,2)*0.4

fig, ax = plt.subplots(1,3,figsize=(12,4))
ax[0].imshow(orig, cmap="gray");     ax[0].set_title("Original");  ax[0].axis("off")
ax[1].imshow(heat_np, cmap="jet");   ax[1].set_title("Grad‑CAM"); ax[1].axis("off")
ax[2].imshow(overlay.astype("uint8")); ax[2].set_title("Overlay"); ax[2].axis("off")
plt.show()
```
